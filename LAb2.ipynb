Сгенерировать с использованием библиотеки pandas csv файл, содержащий 3 столбца:
 (> 400 строк, генерация с использованием функции linspace). Вид функции определяется вариантом (диапазон выбирается студентом).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x1 = np.linspace(-10, 10, 401).flatten()
x2 = np.linspace(-5,5, 401)
y = np.cos(x1) * x2 ** 3
data = {
    "x1": x1,
    "x2": x2,
    "y": y
}
df = pd.DataFrame(data)


df.to_csv('generated_data.csv', index=False)
```

Открыть файл с использованием библиотеки pandas и построить графики:
 Графики необходимо построить с использованием matplotlib (для функции y отобразить точки на графике).

```python
X2mean = df.x2.mean()
y1 = np.cos(x1) * X2mean ** 3
plt.plot(y1, x1, "o")
plt.show
```

```python
X1mean = df.x1.mean()
y2 = np.cos(X1mean) * x2 ** 3
plt.plot(y2, x2, "o")
plt.show
```

Вывести для каждого столбца
: среднее, минимальное и максимальное значения.

```python
print(np.mean(x1), np.min(x1), np.max(x1))
print(np.mean(x2), np.min(x2), np.max(x2))
print(np.mean(y), np.min(y), np.max(y))
```

Сохранить в новый csv файл те строки, для которых выполняется условие:
 меньше среднее
 или
 меньше среднее


```python
filtered_data = df[(df["x1"] < X1mean) | (df["x2"] < X2mean)].copy()
filtered_data.to_csv('filtered_data.csv', index=False)
```

С использованием mplot3D построить 3D график функции
 в отдельном окне.



```python
fig = plt.figure()
fig = fig.add_subplot(projection='3d')
fig.scatter(df['x1'], df['x2'], df['y'], c=df['y'], cmap='inferno', s=20)
plt.show()
```
